# Data Cleaning & Preprocessing

Real-world datasets often contain missing values, incorrect data types, duplicate records, and unnecessary columns.

Before performing Exploratory Data Analysis (EDA), it is important to clean and prepare the data to ensure accurate and meaningful insights.

In this section, we will:

- Remove unnecessary columns
- Convert incorrect data types
- Handle missing values
- Check for duplicates
- Detect and treat outliers

## Create a Working Copy

A copy of the original dataset is created to preserve the raw data. This ensures that the original dataset remains unchanged throughout the analysis.

```python
# Create a copy of the dataset
youtube = df.copy()

youtube.head()
```

## View Column Names

Before cleaning the dataset, it is important to understand the available features.

```python
youtube.columns
```

## Remove Unnecessary Columns

Some columns do not contribute to the analysis because they are unique identifiers or contain information that is not required for understanding video popularity.

Removing such columns simplifies the dataset and improves readability.

```python
# Drop unnecessary columns

youtube.drop(columns=["Video_id"], inplace=True)

youtube.head()
```

## Check Data Types

Some numerical columns may be stored as object (string) data type due to missing values or formatting issues.

These columns must be converted into numeric format before performing statistical analysis.

```python
youtube.dtypes
```

## Identify Numerical and Categorical Features

Separating numerical and categorical variables helps perform appropriate analysis and visualization for each type.

```python
# Numerical columns
num_cols = youtube.select_dtypes(include=np.number).columns
print("Numerical Columns:\n")
print(num_cols)

print("-"*60)

# Categorical columns
cat_cols = youtube.select_dtypes(include="object").columns
print("Categorical Columns:\n")
print(cat_cols)
```

## Convert Data Types

Some columns that represent numerical values are stored as object data types.

These columns are converted into numeric format to enable statistical analysis and visualization.

```python
numeric_columns = [
    "subscriber",
    "views",
    "likes",
    "dislike",
    "comment_count",
    "Trend_day_count",
    "Tag_count",
    "Trend_tag_count"
]

for col in numeric_columns:
    youtube[col] = pd.to_numeric(youtube[col], errors="coerce")
```

```python
youtube.info()
```

## Missing Value Analysis

Missing values can reduce the quality of analysis.

Before handling them, we first identify which columns contain missing values and calculate their percentage.

```python
missing = pd.DataFrame({
    "Missing Values": youtube.isnull().sum(),
    "Percentage": round(youtube.isnull().mean()*100,2)
})

missing[missing["Missing Values"]>0].sort_values("Percentage",ascending=False)
```

### Handling Missing Values in Numerical Features

The median is used because it is less affected by extreme values (outliers) compared to the mean.

```python
for col in youtube.select_dtypes(include=np.number).columns:
    youtube[col].fillna(youtube[col].median(), inplace=True)
```

### Handling Missing Values in Categorical Features

Categorical missing values are replaced using the most frequent value (mode), which preserves the existing distribution of the data.

```python
for col in youtube.select_dtypes(include="object").columns:
    youtube[col].fillna(youtube[col].mode()[0], inplace=True)
```

```python
youtube.isnull().sum()
```

# Check Duplicate Records Again

```python
print("Duplicate Rows:", youtube.duplicated().sum())
```

```python
youtube.drop_duplicates(inplace=True)

print("Duplicate Rows after removal:", youtube.duplicated().sum())
```

# Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of examining, summarizing, and visualizing data to discover meaningful patterns, relationships, and trends.

The objective of this section is to:

- Understand the distribution of numerical variables.
- Identify patterns in audience engagement.
- Compare different video categories and channels.

- Generate meaningful business insights from the data.

Each visualization is accompanied by an explanation, observations, and business insights to demonstrate the analytical reasoning behind the results.

# Analysis 1 : Distribution of Video Views

## Objective

To understand how video views are distributed across trending videos.

This analysis helps determine whether most videos receive similar numbers of views or whether a small number of viral videos dominate the platform.

```python
plt.figure(figsize=(10,6))

sns.histplot(youtube["views"],
             bins=30,
             kde=True,
             color="royalblue")

plt.title("Distribution of Video Views", fontsize=16)

plt.xlabel("Views")

plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()
```

### Observation

- The distribution is positively skewed.
- Most videos receive moderate numbers of views.
- Only a few videos achieve extremely high view counts.

### Business Insight

A small percentage of videos become viral and contribute significantly to overall platform engagement. Content creators should focus on producing high-quality and engaging content that has the potential to reach a wider audience.

# Analysis 5 : Outlier Detection

## Objective

Boxplots help identify extreme values (outliers) in numerical variables.

Outliers are important because they may represent viral videos or unusually successful channels.

```python
plt.figure(figsize=(14,7))

sns.boxplot(data=youtube.select_dtypes(include="number"))

plt.xticks(rotation=90)

plt.title("Boxplot of Numerical Features")

plt.show()
```

# Analysis 6 : Correlation Heatmap

## Objective

To examine the relationships among numerical variables.

Correlation analysis helps identify which variables move together and influence one another.

```python
plt.figure(figsize=(12,8))

corr = youtube.select_dtypes(include="number").corr()

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

- Views show strong positive relationships with likes and comments.
- Likes and comments are also positively correlated.
- Subscriber count has a moderate positive relationship with views.

### Business Insight

Higher audience engagement generally leads to higher video popularity. Creators who encourage likes, comments, and interaction are more likely to achieve greater visibility.

# Analysis 7 : Top 10 Trending Channels

## Objective

This analysis identifies the channels that appear most frequently in the trending videos dataset.

Understanding which channels consistently trend helps identify creators with strong audience engagement and successful content strategies.

```python
# Top 10 Trending Channels

top_channels = youtube["channel_title"].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_channels.values,
    y=top_channels.index,
    palette="viridis"
)

plt.title("Top 10 Trending YouTube Channels", fontsize=16)

plt.xlabel("Number of Trending Videos")

plt.ylabel("Channel Name")

plt.show()
```

# Analysis 14 : Relationship between Views and Likes

## Objective

The objective of this analysis is to examine whether videos with higher views also receive more likes.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=youtube,
    x="views",
    y="likes",
    alpha=0.6,
    color="royalblue"
)

plt.title("Views vs Likes", fontsize=16)

plt.xlabel("Views")

plt.ylabel("Likes")

plt.grid(alpha=0.3)

plt.show()
```

#  Analysis 16 : Likes vs Comments

#  Analysis 20 : Correlation Matrix

## Objective

To measure the strength and direction of relationships between numerical variables.

Correlation values range from -1 to +1.

- +1 → Strong Positive Relationship
- 0 → No Relationship
- -1 → Strong Negative Relationship

```python
plt.figure(figsize=(12,8))

corr = youtube.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="RdYlBu",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Correlation Matrix", fontsize=16)

plt.show()
```

### Observation

- Views have a strong positive correlation with likes and comments.
- Likes and comments are also positively related.
- Subscriber count has a moderate positive relationship with views.

### Business Insight

Videos with higher engagement tend to perform better overall. Encouraging audience interaction can contribute to improved video performance and visibility.
